# CellChat Panel C: significant regional slopes

Creates a dot-and-whisker plot of donor-level pathology slopes for FDR-significant ligand–receptor features in AG, EC, MTC, and TH. Points show the fitted slope and whiskers show the 95% HC3 confidence interval. The TSV beside this notebook contains the plotted estimates and p/FDR values.

Run the cells from top to bottom. Edit the settings in the next cell to change the plot dimensions and font sizes. The plot, table, and notebook are kept together in this folder.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt

HERE = Path.cwd()
if not (HERE / 'panel_c_dotplot.py').exists():
    HERE = Path(__file__).resolve().parent if '__file__' in globals() else HERE
if str(HERE) not in sys.path:
    sys.path.insert(0, str(HERE))
from panel_c_dotplot import REGIONS, find_project_root, load_significant_results, plot_dot_whisker, save_plot_table

# Adjustable plot settings
REGIONS_TO_PLOT = ['AG', 'EC', 'MTC', 'TH']
FIGURE_WIDTH_INCHES = 5
ROW_HEIGHT_INCHES = 0.38
LABEL_FONTSIZE = 16
TICK_FONTSIZE = 16
AXIS_FONTSIZE = 28
TITLE_FONTSIZE = 28
P_FDR_FONTSIZE = 16
PNG_DPI = 300
OUTPUT_DIR = HERE
PROJECT_ROOT = find_project_root()
print(f'Project: {PROJECT_ROOT}')
print(f'Outputs: {OUTPUT_DIR}')
```

```python
results = load_significant_results(PROJECT_ROOT, regions=REGIONS_TO_PLOT)
table_path = save_plot_table(results, OUTPUT_DIR)
print(f'{len(results)} FDR-significant features')
print(results.groupby('region').size().reindex(REGIONS_TO_PLOT, fill_value=0).to_string())
print(f'Plot table: {table_path}')
display(results[['region', 'source', 'target', 'interaction_id', 'beta_pathology_stage', 'ci_low', 'ci_high', 'p_value', 'q_value_within_family']].head())
```

```python
for region in REGIONS_TO_PLOT:
    plot_dot_whisker(
        results, OUTPUT_DIR, region=region,
        width=FIGURE_WIDTH_INCHES, row_height=ROW_HEIGHT_INCHES,
        label_fontsize=LABEL_FONTSIZE, tick_fontsize=TICK_FONTSIZE,
        axis_fontsize=AXIS_FONTSIZE, title_fontsize=TITLE_FONTSIZE,
        stats_fontsize=P_FDR_FONTSIZE, dpi=PNG_DPI,
    )
print('Saved one PDF and PNG dot plot per region in', OUTPUT_DIR)
```
